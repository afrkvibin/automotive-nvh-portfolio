```python
import numpy as np
import matplotlib.pyplot as plt

fs = 1000 # Sampling rate, Hz
f_true = 700 # real signal frequency, Hz
duration = 0.01 # 10 ms 
```

```python
t_fine = np.linspace(0, duration, 10000, endpoint=False)
x_true = np.sin(2*np.pi*f_true*t_fine)

n = np.arange(int(fs*duration))   # sample numbers 0, 1, 2 ...
t_s = n / fs                      # sample times, 1 ms apart
x_s = np.sin(2*np.pi*f_true*t_s)
```

```python
f_alias = abs(f_true - round(f_true/fs)*fs)
print("Alias frequency:", f_alias, "Hz")
x_alias = -np.sin(2*np.pi*f_alias*t_fine)
```

```python
plt.figure(figsize=(10, 4))
plt.plot(t_fine*1000, x_true, alpha=0.4, label="True 700 Hz")
plt.plot(t_fine*1000, x_alias, "--", label=f"Alias {f_alias} Hz")
plt.plot(t_s*1000, x_s, "ko", label="Samples at 1000 Hz")
plt.xlabel("Time [ms]"); plt.ylabel("Amplitude [-]")
plt.title("Aliasing: 700 Hz sampled at 1000 Hz")
plt.legend(); plt.grid(True); plt.show()
```

With the Sampling rate adjusted to 2000 Hz, the Nyquist fequency becomes 1000 Hz which is above 700 Hz, and that is why the dots are sitting on the the 700 Hz line
